**1. Import Libraries**

```python
from transformers import BertTokenizer, BertModel
import torch
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
```

**2. Load the pre-trained BERT tokenizer and model**

```python
tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")
bert_model = BertModel.from_pretrained("bert-base-uncased")
```

**3. Example sentence pairs**

```python
sentence_pairs = [
    ("How do I learn Python?", "What is the best way to study Python?"),
    ("What is AI?", "How to cook pasta?"),
    ("How do I bake a chocolate cake?", "Give me a chocolate cake recipe."),
    ("How can I improve my coding skills?", "Tips for becoming better at programming."),
    ("Where can I buy cheap laptops?", "Best sites to find affordable computers."),

    # 5 new pairs
    ("The bank will close at 4 PM.", "I need to deposit money before the bank closes."),
    ("The bank of the river was steep.", "I deposited cash at the bank."),
    ("A dog is running in the park.", "A cat is sleeping on the sofa."),
    ("The smartphone battery dies quickly.", "My phone does not last long on a charge."),
    ("He scored a goal in the football match.", "She wrote an essay for her homework.")
]
```

**4. Ground truth similarity labels**

```python
# 1 = similar 0 = not similar
labels = [1, 0, 1, 1, 1,
          1, 0, 0, 1, 0]
```

**5. Function to Extract BERT 'CLS' Sentence Embeddings**

```python
def get_sentence_embedding(sentence):
    inputs = tokenizer(sentence, return_tensors="pt", padding=True, truncation=True)
    outputs = bert_model(**inputs)
    cls_embedding = outputs.last_hidden_state[:, 0, :]
    return cls_embedding.detach().numpy()
```

**6. Compute cosine similarity for each pair**

```python
predictions = []

for sent1, sent2 in sentence_pairs:
    emb1 = get_sentence_embedding(sent1)
    emb2 = get_sentence_embedding(sent2)

    sim_score = cosine_similarity(emb1, emb2)[0][0]

    # Classification threshold
    pred = 1 if sim_score > 0.7 else 0
    predictions.append(pred)

    print(f"\nSentence 1: {sent1}")
    print(f"Sentence 2: {sent2}")
    print(f"Cosine Similarity: {sim_score:.4f} → Predicted: {pred}")
```

**7. Evaluate accuracy and final accuracy calculation**

```python
correct = sum([1 for p, y in zip(predictions, labels) if p == y])
accuracy = correct / len(labels)

print("\nFinal Accuracy:", round(accuracy * 100, 2), "%")
```
